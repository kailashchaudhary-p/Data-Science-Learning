```python
import seaborn as sns
import matplotlib.pyplot as plt
# Example dataset
tips = sns.load_dataset("tips")
# Create a scatter plot with regression line
sns.lmplot(data=tips, x="total_bill", y="tip", hue="sex", height=5, aspect=1.2)
plt.title("Total Bill vs Tip by Gender")
plt.show()
```

```python
penguins=sns.load_dataset('penguins')
penguins
```

```python
penguins['species'].value_counts()
```

```python
penguins['island'].value_counts()
```

```python
penguins['bill_depth_mm'].value_counts()
```

```python
sns.set_style('white')
sns.set_context('talk')
sns.scatterplot(data=penguins,x='species',y='island')
```

```python
sns.set_style('ticks')
sns.scatterplot(data=penguins,y='bill_depth_mm',x='island',hue='bill_length_mm',style='sex',alpha=1)
```

```python
sns.stripplot(data=penguins,x='species',y='bill_length_mm',hue='sex',dodge=True,jitter=True)
```
