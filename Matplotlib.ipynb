```python
import matplotlib.pyplot as plt

# Data
categories = ["a", "b", "c", "d"]
values = [15, 30, 45, 30]

# Figure size
plt.figure(figsize=(8, 4))

# Bar Plot
plt.subplot(1, 2, 1)
plt.bar(categories, values, color="green", width=0.6)
plt.title("Bar Plot")
plt.xlabel("Categories")
plt.ylabel("Values")

# Line Plot
plt.subplot(1, 2, 2)
plt.plot(categories, values, marker="o", color="blue", linewidth=2)
plt.title("Line Plot")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.grid(True)

# Adjust layout
plt.tight_layout()

# Show plot
plt.show()
```
