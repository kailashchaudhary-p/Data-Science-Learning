```python
import matplotlib.pyplot as plt
#this is the my first chart
x = [10,20,35,40,45,50,55]
y = [15,20,25,30,35,40,45]
plt.plot(x,y)
```

```python
from matplotlib.lines import lineStyles
#customize chart
x=[10,20,35,40,45,50,55]
y1=[15,20,25,30,35,40,45]

plt.plot(x,y1,color="red",marker="2",linestyle="dashed",linewidth=2,markersize=10)
plt.show()
```

```python
#customize chart
x=[105,220,350,408,145,520,575]
y1=[15,20,25,30,35,40,45]

plt.plot(x,y1,color="blue",marker="2",linestyle="dashed",linewidth=2,markersize=10)
plt.show()
```

```python
#advance line chart
x=[105,220,350,408,145,520,575]
y1=[15,20,25,30,35,40,45]
y2=[15,20,25,30,35,40,45]
plt.plot(x,y1,color="blue",marker="2",linestyle="dashed",linewidth=2,markersize=10)
plt.plot(x,y2,color="red",marker="2",linestyle="dashed",linewidth=2,markersize=10)
plt.title("advance line chart")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#most important chart (daily use)
#1) Bar chart
x=[15,20,30,48,45,50,75]
y1=[15,20,25,30,35,40,45]
plt.bar(x,y1)
plt.title("bar chart ")
plt.xlabel("x-axis")
plt.ylabel('y-axis')
plt.show()
```

```python
#2) Histogram
import random
data =[random.randint(0,50) for i in range(100)]
plt.hist(data)
plt.title("histogram")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
#3) pie chart
# data
products=['chips','biscut','candy','pepsi','HELL','Bikaneri bhujia']
sales=[101,55,111,50,60,85]
plt.pie(sales,labels=products)
plt.title('Weekly sales of shop')
plt.show()
```

```python
#4) scaterplot
x=[15,20,30,48,45,50,75]
y1=[15,20,25,30,35,40,45]
plt.scatter(x,y1)
plt.title("scatter chart ")
plt.xlabel("x-axis")
plt.ylabel('y-axis')
plt.show()
```

```python
#5) sub plots
#use = in this we will use the multipple chart in one figur
#data 1
days =['monday','tuesday','wednesday','thursday','friday','saturday','sunday']
sales =[1200,500,600,500,1500,900,700]
#data 2
months=['january','february','march','aprial','may','june','july','august','September','October','November','December']
sales_month=[15000,12000,13000,5000,7000,9000,7000,14000,21000,8000,13000,19000]
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.bar(days,sales)
plt.subplot(1,2,2)
plt.plot(months,sales_month)
plt.show()
```

```python
#6) Matplotlib with Pandas - real data
#first we will creat the data frame
import pandas as pd
my_dict={'product':['Monitor','CPU','Monitor','RAM','Monitor','Monitor','RAM','RAM','Monitor'],
         'qty':[2,1,3,2,3,2,3,2,2],}
sales = pd.DataFrame(data=my_dict)
#use the matplotlib
plt.bar(sales['product'],sales['qty'])
plt.title("Sale of shop")
plt.xlabel("The product")
plt.ylabel("sales")
plt.show()
#for save the graph
plt.savefig('sales_of_shop.png')
```
